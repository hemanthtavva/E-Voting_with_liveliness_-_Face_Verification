## Step 3: Data Preprocessing\n\n### Step 3a: Explore Dataset Structure

```python
from pathlib import Path

def explore_dataset(path, max_depth=2):
    """Explore the dataset directory structure"""
    path = Path(path)
    if not path.exists():
        print(f"❌ Path does not exist: {path}")
        return
    
    print(f"\n📁 Dataset: {path.name}")
    print("=" * 60)
    
    def show_tree(p, prefix="", depth=0):
        if depth >= max_depth:
            return
        try:
            items = sorted(p.iterdir()) if p.is_dir() else []
        except PermissionError:
            return
        for i, item in enumerate(items[:10]):
            is_last = i == len(items[:10]) - 1
            connector = "└── " if is_last else "├── "
            if item.is_dir():
                count = len(list(item.glob('*')))
                print(f"{prefix}{connector}📂 {item.name}/ ({count} items)")
                new_prefix = prefix + ("    " if is_last else "│   ")
                show_tree(item, new_prefix, depth + 1)
            else:
                size = item.stat().st_size / 1024
                print(f"{prefix}{connector}📄 {item.name} ({size:.1f} KB)")
        if len(items) > 10:
            print(f"{prefix}    ... and {len(items) - 10} more items")
    
    show_tree(path)

# Explore both datasets
print("\\n" + "=" * 60)
print("EXPLORING DOWNLOADED DATASETS")
print("=" * 60)

explore_dataset(OULU_NPU_PATH)
explore_dataset(TAPAKAH_PATH)
```

### Step 3b: Create Dataset Class\n\nBased on the dataset structure:\n- **OULU-NPU**: `true/` (real) and `false/` (spoof) folders with images\n- **tapakah68**: `live_selfie/`, `live_video/` (real) and `printouts/`, `replay/`, `cut-out printouts/` (spoof) with videos/images

```python
import cv2
import numpy as np
from torch.utils.data import Dataset
from pathlib import Path

class LivenessDataset(Dataset):
    """
    Dataset for Liveness Detection
    
    Handles both OULU-NPU and tapakah68 dataset structures.
    For videos, extracts frames automatically.
    """
    
    def __init__(self, root_dir, transform=None, max_samples_per_class=None, frames_per_video=5):
        self.root_dir = Path(root_dir)
        self.transform = transform
        self.frames_per_video = frames_per_video
        self.samples = []  # List of (path, label, is_video)
        self.labels = []
        
        # Real folder names (label = 1)
        real_dirs = ['true', 'True', 'real', 'Real', 'live', 'Live', 'live_selfie', 'live_video', 'genuine']
        
        # Spoof folder names (label = 0)
        spoof_dirs = ['false', 'False', 'spoof', 'Spoof', 'fake', 'Fake', 'attack', 
                      'printouts', 'replay', 'cut-out printouts']
        
        # Search for real images/videos
        for dir_name in real_dirs:
            real_path = self._find_dir(dir_name)
            if real_path:
                self._load_files(real_path, label=1, max_samples=max_samples_per_class)
        
        # Search for spoof images/videos
        for dir_name in spoof_dirs:
            spoof_path = self._find_dir(dir_name)
            if spoof_path:
                self._load_files(spoof_path, label=0, max_samples=max_samples_per_class)
        
        print(f"\\nDataset loaded from: {root_dir}")
        print(f"  Total samples: {len(self.samples)}")
        print(f"  Real: {sum(self.labels)}")
        print(f"  Spoof: {len(self.labels) - sum(self.labels)}")
    
    def _find_dir(self, dir_name):
        """Find directory by name (searches recursively)"""
        for path in self.root_dir.rglob(dir_name):
            if path.is_dir():
                return path
        return None
    
    def _load_files(self, path, label, max_samples=None):
        """Load images and videos from directory"""
        image_exts = ['.jpg', '.jpeg', '.png', '.bmp']
        video_exts = ['.mp4', '.avi', '.mov', '.MOV']
        
        files = list(path.glob('*'))
        count = 0
        
        for file_path in files:
            if max_samples and count >= max_samples:
                break
            
            ext = file_path.suffix.lower()
            
            if ext in image_exts:
                self.samples.append((str(file_path), label, False))  # is_video = False
                self.labels.append(label)
                count += 1
            elif ext in video_exts or file_path.suffix in video_exts:
                self.samples.append((str(file_path), label, True))  # is_video = True
                self.labels.append(label)
                count += 1
    
    def _extract_frame_from_video(self, video_path):
        """Extract a random frame from video"""
        cap = cv2.VideoCapture(video_path)
        if not cap.isOpened():
            return np.zeros((224, 224, 3), dtype=np.uint8)
        
        total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
        if total_frames <= 0:
            cap.release()
            return np.zeros((224, 224, 3), dtype=np.uint8)
        
        # Get random frame
        frame_idx = np.random.randint(0, total_frames)
        cap.set(cv2.CAP_PROP_POS_FRAMES, frame_idx)
        ret, frame = cap.read()
        cap.release()
        
        if not ret or frame is None:
            return np.zeros((224, 224, 3), dtype=np.uint8)
        
        return cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    
    def __len__(self):
        return len(self.samples)
    
    def __getitem__(self, idx):
        file_path, label, is_video = self.samples[idx]
        
        # Load image or extract frame from video
        if is_video:
            image = self._extract_frame_from_video(file_path)
        else:
            image = cv2.imread(file_path)
            if image is None:
                image = np.zeros((224, 224, 3), dtype=np.uint8)
            else:
                image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        
        # Apply transforms
        if self.transform:
            image = self.transform(image)
        
        return image, label

print("✅ LivenessDataset class defined")
```

### Step 3c: Define Transforms\n\nData augmentation for training and normalization for validation.

```python
from torchvision import transforms

# Training transforms with data augmentation
train_transform = transforms.Compose([
    transforms.ToPILImage(),
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
    transforms.RandomRotation(15),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# Validation transforms (no augmentation)
val_transform = transforms.Compose([
    transforms.ToPILImage(),
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

print("✅ Transforms defined")
print("   - Train: Resize, Flip, ColorJitter, Rotation, Normalize")
print("   - Val: Resize, Normalize")
```

### Step 3d: Create Dataloaders\n\nCreate train and validation dataloaders for both datasets.

```python
import torch
from torch.utils.data import DataLoader, random_split

# Set device - Use MPS (Apple Silicon GPU) if available
if torch.backends.mps.is_available():
    device = torch.device('mps')
    print("✅ Using Apple Silicon GPU (MPS)")
elif torch.cuda.is_available():
    device = torch.device('cuda')
    print("✅ Using NVIDIA GPU (CUDA)")
else:
    device = torch.device('cpu')
    print("⚠️ Using CPU (slower)")

print(f"Device: {device}")

# Create datasets
print("\n" + "=" * 60)
print("LOADING DATASETS")
print("=" * 60)

# Load tapakah68 dataset (for pre-training)
print("\n📂 Loading tapakah68 dataset...")
tapakah_dataset = LivenessDataset(TAPAKAH_PATH, transform=train_transform)

# Load OULU-NPU dataset (for fine-tuning)
print("\n📂 Loading OULU-NPU dataset...")
oulu_dataset = LivenessDataset(OULU_NPU_PATH, transform=train_transform)

# Split tapakah68 into train/val (80/20)
tapakah_train_size = int(0.8 * len(tapakah_dataset))
tapakah_val_size = len(tapakah_dataset) - tapakah_train_size
tapakah_train, tapakah_val = random_split(tapakah_dataset, [tapakah_train_size, tapakah_val_size])

# Split OULU-NPU into train/val (80/20)
oulu_train_size = int(0.8 * len(oulu_dataset))
oulu_val_size = len(oulu_dataset) - oulu_train_size
oulu_train, oulu_val = random_split(oulu_dataset, [oulu_train_size, oulu_val_size])

# Create dataloaders
BATCH_SIZE = 32

tapakah_train_loader = DataLoader(tapakah_train, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
tapakah_val_loader = DataLoader(tapakah_val, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

oulu_train_loader = DataLoader(oulu_train, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
oulu_val_loader = DataLoader(oulu_val, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

print("\n" + "=" * 60)
print("DATALOADERS READY")
print("=" * 60)
print(f"\ntapakah68:")
print(f"  Train: {len(tapakah_train)} samples, {len(tapakah_train_loader)} batches")
print(f"  Val:   {len(tapakah_val)} samples, {len(tapakah_val_loader)} batches")
print(f"\nOULU-NPU:")
print(f"  Train: {len(oulu_train)} samples, {len(oulu_train_loader)} batches")
print(f"  Val:   {len(oulu_val)} samples, {len(oulu_val_loader)} batches")
print("=" * 60)
```

## Step 4: Model Architecture\n\nMobileNetV2 backbone with custom classifier head for binary classification (Real vs Spoof).

```python
import torch.nn as nn
from torchvision import models

class LivenessDetector(nn.Module):
    """
    Liveness Detection Model using MobileNetV2 backbone
    
    Architecture:
    - MobileNetV2 backbone (pretrained on ImageNet)
    - Custom classifier head for binary classification
    """
    
    def __init__(self, pretrained=True, dropout=0.5):
        super().__init__()
        
        # Load MobileNetV2 backbone
        self.backbone = models.mobilenet_v2(weights='IMAGENET1K_V1' if pretrained else None)
        
        # Get feature dimension from the last layer
        self.feature_dim = self.backbone.classifier[1].in_features
        
        # Replace classifier with custom head
        self.backbone.classifier = nn.Sequential(
            nn.Dropout(dropout),
            nn.Linear(self.feature_dim, 256),
            nn.ReLU(inplace=True),
            nn.Dropout(dropout / 2),
            nn.Linear(256, 2)  # [spoof, real]
        )
        
        print(f"✅ LivenessDetector initialized")
        print(f"   Backbone: MobileNetV2")
        print(f"   Feature dim: {self.feature_dim}")
        print(f"   Pretrained: {pretrained}")
    
    def forward(self, x):
        return self.backbone(x)
    
    def predict(self, x):
        """Get prediction probabilities"""
        self.eval()
        with torch.no_grad():
            logits = self.forward(x)
            probs = torch.softmax(logits, dim=1)
            return probs[:, 1]  # Return probability of being real
    
    def freeze_backbone(self):
        """Freeze backbone for transfer learning"""
        for name, param in self.backbone.named_parameters():
            if 'classifier' not in name:
                param.requires_grad = False
        print("🔒 Backbone frozen (only classifier trainable)")
    
    def unfreeze_backbone(self):
        """Unfreeze backbone for fine-tuning"""
        for param in self.backbone.parameters():
            param.requires_grad = True
        print("🔓 Backbone unfrozen (all layers trainable)")

# Create model
model = LivenessDetector(pretrained=True)
model = model.to(device)

# Count parameters
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

print(f"\n📊 Model Statistics:")
print(f"   Total parameters: {total_params:,}")
print(f"   Trainable parameters: {trainable_params:,}")
```

## Step 5: Training\n\n### Stage 1: Pre-train on OULU-NPU dataset (larger dataset)\n\nSince OULU-NPU has more samples (3402) compared to tapakah68 (45), we'll train on OULU-NPU first.

```python
import torch.optim as optim
from tqdm import tqdm

def train_epoch(model, train_loader, criterion, optimizer, device):
    """Train for one epoch"""
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0
    
    pbar = tqdm(train_loader, desc='Training')
    for images, labels in pbar:
        images = images.to(device)
        labels = labels.to(device)
        
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        
        running_loss += loss.item()
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()
        
        pbar.set_postfix({'loss': f'{loss.item():.4f}', 'acc': f'{100.*correct/total:.2f}%'})
    
    return running_loss / len(train_loader), 100. * correct / total

def validate(model, val_loader, criterion, device):
    """Validate the model"""
    model.eval()
    running_loss = 0.0
    correct = 0
    total = 0
    
    with torch.no_grad():
        pbar = tqdm(val_loader, desc='Validation')
        for images, labels in pbar:
            images = images.to(device)
            labels = labels.to(device)
            
            outputs = model(images)
            loss = criterion(outputs, labels)
            
            running_loss += loss.item()
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()
            
            pbar.set_postfix({'loss': f'{loss.item():.4f}', 'acc': f'{100.*correct/total:.2f}%'})
    
    return running_loss / len(val_loader), 100. * correct / total

def train_model(model, train_loader, val_loader, num_epochs, lr, device, save_path):
    """Full training loop"""
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=lr, weight_decay=1e-4)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=3)
    
    best_val_acc = 0.0
    history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}
    
    print(f"\n{'='*60}")
    print(f"TRAINING")
    print(f"{'='*60}")
    print(f"Epochs: {num_epochs}")
    print(f"Learning Rate: {lr}")
    print(f"Device: {device}")
    print(f"{'='*60}\n")
    
    for epoch in range(num_epochs):
        print(f"\nEpoch {epoch+1}/{num_epochs}")
        print("-" * 40)
        
        train_loss, train_acc = train_epoch(model, train_loader, criterion, optimizer, device)
        val_loss, val_acc = validate(model, val_loader, criterion, device)
        
        scheduler.step(val_loss)
        
        history['train_loss'].append(train_loss)
        history['train_acc'].append(train_acc)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)
        
        print(f"\n  Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}%")
        print(f"  Val Loss:   {val_loss:.4f} | Val Acc:   {val_acc:.2f}%")
        
        if val_acc > best_val_acc:
            best_val_acc = val_acc
            torch.save({
                'epoch': epoch,
                'model_state_dict': model.state_dict(),
                'optimizer_state_dict': optimizer.state_dict(),
                'val_acc': val_acc,
                'val_loss': val_loss,
            }, save_path)
            print(f"  ✅ Saved best model (acc: {val_acc:.2f}%)")
    
    print(f"\n{'='*60}")
    print(f"TRAINING COMPLETE")
    print(f"{'='*60}")
    print(f"Best Validation Accuracy: {best_val_acc:.2f}%")
    print(f"Model saved to: {save_path}")
    print(f"{'='*60}\n")
    
    return history

print("✅ Training functions defined")
```

```python
# Create models directory
import os
os.makedirs('../models', exist_ok=True)

# Freeze backbone for initial training (train only classifier)
model.freeze_backbone()

# Train on OULU-NPU dataset (Stage 1)
NUM_EPOCHS = 10
LEARNING_RATE = 0.001

history = train_model(
    model=model,
    train_loader=oulu_train_loader,
    val_loader=oulu_val_loader,
    num_epochs=NUM_EPOCHS,
    lr=LEARNING_RATE,
    device=device,
    save_path='../models/liveness_stage1.pth'
)
```

### Stage 2: Fine-tune with unfrozen backbone\n\nUnfreeze the backbone and fine-tune with a lower learning rate.

```python
# Load best model from Stage 1
checkpoint = torch.load('../models/liveness_stage1.pth')
model.load_state_dict(checkpoint['model_state_dict'])
print(f"✅ Loaded Stage 1 model (acc: {checkpoint['val_acc']:.2f}%)")

# Unfreeze backbone for fine-tuning
model.unfreeze_backbone()

# Fine-tune with lower learning rate
FINETUNE_EPOCHS = 5
FINETUNE_LR = 0.0001

finetune_history = train_model(
    model=model,
    train_loader=oulu_train_loader,
    val_loader=oulu_val_loader,
    num_epochs=FINETUNE_EPOCHS,
    lr=FINETUNE_LR,
    device=device,
    save_path='../models/liveness_final.pth'
)
```

## Step 6: Evaluation and Export\n\n### 6a: Evaluate Final Model

```python
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Load the best model
checkpoint = torch.load('../models/liveness_final.pth')
model.load_state_dict(checkpoint['model_state_dict'])
model.eval()
print(f"✅ Loaded final model (acc: {checkpoint['val_acc']:.2f}%)")

# Evaluate on validation set
all_preds = []
all_labels = []

with torch.no_grad():
    for images, labels in tqdm(oulu_val_loader, desc='Evaluating'):
        images = images.to(device)
        outputs = model(images)
        _, predicted = outputs.max(1)
        all_preds.extend(predicted.cpu().numpy())
        all_labels.extend(labels.numpy())

# Classification Report
print("\n" + "=" * 60)
print("CLASSIFICATION REPORT")
print("=" * 60)
print(classification_report(all_labels, all_preds, target_names=['Spoof', 'Real']))

# Confusion Matrix
cm = confusion_matrix(all_labels, all_preds)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['Spoof', 'Real'], 
            yticklabels=['Spoof', 'Real'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix - Liveness Detection')
plt.tight_layout()
plt.savefig('../models/confusion_matrix.png', dpi=150)
plt.show()

print(f"\n✅ Confusion matrix saved to: ../models/confusion_matrix.png")
```

### 6c: Test Inference\n\nQuick test to verify the model works correctly.

```python
def predict_liveness(image_path, model, device):
    """
    Predict if an image is real or spoof
    
    Args:
        image_path: Path to image file
        model: Trained LivenessDetector model
        device: torch device
    
    Returns:
        dict with prediction, confidence, and label
    """
    # Load and preprocess image
    image = cv2.imread(image_path)
    if image is None:
        return {"error": "Could not load image"}
    
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    
    # Apply validation transform
    input_tensor = val_transform(image).unsqueeze(0).to(device)
    
    # Predict
    model.eval()
    with torch.no_grad():
        outputs = model(input_tensor)
        probs = torch.softmax(outputs, dim=1)
        pred_class = outputs.argmax(dim=1).item()
        confidence = probs[0][pred_class].item()
    
    labels = ['SPOOF', 'REAL']
    
    return {
        'prediction': labels[pred_class],
        'confidence': confidence * 100,
        'is_real': pred_class == 1,
        'spoof_prob': probs[0][0].item() * 100,
        'real_prob': probs[0][1].item() * 100
    }

# Test on a sample image from validation set
test_path, test_label, _ = oulu_dataset.samples[0]
result = predict_liveness(test_path, model, device)

print("=" * 60)
print("INFERENCE TEST")
print("=" * 60)
print(f"\nTest image: {test_path.split('/')[-1]}")
print(f"Actual label: {'REAL' if test_label == 1 else 'SPOOF'}")
print(f"\nPrediction: {result['prediction']}")
print(f"Confidence: {result['confidence']:.2f}%")
print(f"  - Real probability:  {result['real_prob']:.2f}%")
print(f"  - Spoof probability: {result['spoof_prob']:.2f}%")
print("=" * 60)
```

### Fix: Adjusted Threshold for Real-World Use\n\nThe model is biased toward "Spoof" due to training data imbalance. We'll lower the threshold for classifying as "Real".

```python
import cv2
import time

def test_liveness_webcam_adjusted(model, device, duration=30, threshold=0.3):
    """
    Test liveness detection with webcam (with adjusted threshold)
    
    Args:
        model: Trained LivenessDetector model
        device: torch device
        duration: How long to run the test (seconds)
        threshold: Probability threshold for "Real" classification (lowered due to bias)
    """
    cap = cv2.VideoCapture(0)
    if not cap.isOpened():
        print("❌ Could not open webcam")
        return
    
    print("=" * 60)
    print("LIVE WEBCAM LIVENESS DETECTION (ADJUSTED THRESHOLD)")
    print("=" * 60)
    print(f"Threshold for REAL: {threshold*100:.0f}%")
    print(f"Running for {duration} seconds...")
    print("Press 'q' to quit early")
    print("=" * 60)
    
    model.eval()
    start_time = time.time()
    frame_count = 0
    
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        
        # Preprocess frame
        rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        input_tensor = val_transform(rgb_frame).unsqueeze(0).to(device)
        
        # Predict
        with torch.no_grad():
            outputs = model(input_tensor)
            probs = torch.softmax(outputs, dim=1)
            real_prob = probs[0][1].item()
            spoof_prob = probs[0][0].item()
        
        # Use adjusted threshold
        is_real = real_prob >= threshold
        label = "REAL" if is_real else "SPOOF"
        color = (0, 255, 0) if is_real else (0, 0, 255)
        confidence = real_prob if is_real else spoof_prob
        
        # Add text overlay
        cv2.putText(frame, f"{label}: {confidence*100:.1f}%", (10, 30), 
                    cv2.FONT_HERSHEY_SIMPLEX, 1, color, 2)
        cv2.putText(frame, f"Real: {real_prob*100:.1f}%", (10, 70),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
        cv2.putText(frame, f"Spoof: {spoof_prob*100:.1f}%", (10, 100),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
        cv2.putText(frame, f"Threshold: {threshold*100:.0f}%", (10, 130),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (200, 200, 200), 1)
        
        # Show frame
        cv2.imshow('Liveness Detection (Adjusted)', frame)
        frame_count += 1
        
        # Check for quit or timeout
        if cv2.waitKey(1) & 0xFF == ord('q'):
            print("\n⏹️ Stopped by user")
            break
        if time.time() - start_time > duration:
            print(f"\n⏱️ Test completed ({duration}s)")
            break
    
    cap.release()
    cv2.destroyAllWindows()
    
    elapsed = time.time() - start_time
    fps = frame_count / elapsed
    print(f"\nProcessed {frame_count} frames in {elapsed:.1f}s ({fps:.1f} FPS)")
    print("=" * 60)

# Run with adjusted threshold (lowered from 0.5 to 0.3)
print("Starting webcam test with ADJUSTED threshold...")
print("Threshold lowered to 30% for Real classification")
print("\nTry showing your face (REAL) and then a photo on your phone (SPOOF)")
time.sleep(3)
test_liveness_webcam_adjusted(model, device, duration=30, threshold=0.15)
```

## 🔧 Fix: Retrain with Balanced Dataset\n\nThe model is biased because OULU-NPU has 4x more spoof than real samples. Let's retrain with balanced classes.

```python
# Create BALANCED dataset by undersampling spoof class
from torch.utils.data import Subset
import random

# Get indices for real and spoof samples
real_indices = [i for i, (_, label, _) in enumerate(oulu_dataset.samples) if label == 1]
spoof_indices = [i for i, (_, label, _) in enumerate(oulu_dataset.samples) if label == 0]

print(f"Original dataset:")
print(f"  Real samples: {len(real_indices)}")
print(f"  Spoof samples: {len(spoof_indices)}")

# Undersample spoof to match real
random.seed(42)
balanced_spoof_indices = random.sample(spoof_indices, len(real_indices))
balanced_indices = real_indices + balanced_spoof_indices
random.shuffle(balanced_indices)

print(f"\nBalanced dataset:")
print(f"  Real samples: {len(real_indices)}")
print(f"  Spoof samples: {len(balanced_spoof_indices)}")
print(f"  Total: {len(balanced_indices)}")

# Create balanced dataset
balanced_dataset = Subset(oulu_dataset, balanced_indices)

# Split into train/val (80/20)
balanced_train_size = int(0.8 * len(balanced_dataset))
balanced_val_size = len(balanced_dataset) - balanced_train_size
balanced_train, balanced_val = random_split(balanced_dataset, [balanced_train_size, balanced_val_size])

# Create dataloaders
balanced_train_loader = DataLoader(balanced_train, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
balanced_val_loader = DataLoader(balanced_val, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

print(f"\nBalanced DataLoaders:")
print(f"  Train: {len(balanced_train)} samples, {len(balanced_train_loader)} batches")
print(f"  Val: {len(balanced_val)} samples, {len(balanced_val_loader)} batches")
```

```python
# Create a NEW model and retrain with balanced data
model_balanced = LivenessDetector(pretrained=True)
model_balanced = model_balanced.to(device)

# Freeze backbone first
model_balanced.freeze_backbone()

# Train on balanced dataset (Stage 1)
print("\n" + "=" * 60)
print("RETRAINING WITH BALANCED DATA")
print("=" * 60)

balanced_history = train_model(
    model=model_balanced,
    train_loader=balanced_train_loader,
    val_loader=balanced_val_loader,
    num_epochs=10,
    lr=0.001,
    device=device,
    save_path='../models/liveness_balanced_stage1.pth'
)
```

## 🔧 Proper Approach: Face Detection + Liveness\n\nThe model was trained on **cropped face images**, not full frames. We need to:\n1. Detect face using MTCNN\n2. Crop the face region\n3. Run liveness detection on the cropped face

```python
# Use OpenCV's DNN face detector instead of MTCNN (no extra dependencies needed)
# Download the pre-trained face detector model
import urllib.request
import os

# Create directory for models
os.makedirs('../models/face_detector', exist_ok=True)

# Download OpenCV DNN face detector files
prototxt_url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/dnn/face_detector/deploy.prototxt"
caffemodel_url = "https://raw.githubusercontent.com/opencv/opencv_3rdparty/dnn_samples_face_detector_20170830/res10_300x300_ssd_iter_140000.caffemodel"

prototxt_path = '../models/face_detector/deploy.prototxt'
caffemodel_path = '../models/face_detector/res10_300x300_ssd_iter_140000.caffemodel'

if not os.path.exists(prototxt_path):
    print("Downloading face detector prototxt...")
    urllib.request.urlretrieve(prototxt_url, prototxt_path)
    print("✅ Downloaded deploy.prototxt")

if not os.path.exists(caffemodel_path):
    print("Downloading face detector model (10MB)...")
    urllib.request.urlretrieve(caffemodel_url, caffemodel_path)
    print("✅ Downloaded caffemodel")

print("✅ Face detector files ready!")
```

```python
# Proper Pipeline: Face Detection (OpenCV DNN) + Liveness Detection
import cv2
import time

# Load OpenCV DNN face detector
face_net = cv2.dnn.readNetFromCaffe(
    '../models/face_detector/deploy.prototxt',
    '../models/face_detector/res10_300x300_ssd_iter_140000.caffemodel'
)
print("✅ OpenCV DNN Face Detector loaded")

def detect_face_opencv(frame, confidence_threshold=0.5):
    """Detect face using OpenCV DNN"""
    h, w = frame.shape[:2]
    blob = cv2.dnn.blobFromImage(frame, 1.0, (300, 300), (104.0, 177.0, 123.0))
    face_net.setInput(blob)
    detections = face_net.forward()
    
    for i in range(detections.shape[2]):
        confidence = detections[0, 0, i, 2]
        if confidence > confidence_threshold:
            box = detections[0, 0, i, 3:7] * np.array([w, h, w, h])
            x1, y1, x2, y2 = box.astype(int)
            return (x1, y1, x2, y2), confidence
    return None, 0

def test_liveness_with_face_detection(model, device, duration=30):
    """
    Proper liveness detection pipeline:
    1. Detect face using OpenCV DNN
    2. Crop face region
    3. Run liveness model on cropped face
    """
    cap = cv2.VideoCapture(0)
    if not cap.isOpened():
        print("❌ Could not open webcam")
        return
    
    print("=" * 60)
    print("LIVENESS DETECTION WITH FACE DETECTION")
    print("=" * 60)
    print("Pipeline: Frame → Face Detection → Crop → Liveness Model")
    print(f"Running for {duration} seconds...")
    print("Press 'q' to quit early")
    print("=" * 60)
    
    model.eval()
    start_time = time.time()
    frame_count = 0
    face_count = 0
    
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        
        # Detect face
        face_box, face_conf = detect_face_opencv(frame)
        
        if face_box is not None:
            face_count += 1
            x1, y1, x2, y2 = face_box
            
            # Add padding (20%)
            h, w = frame.shape[:2]
            pad_x = int((x2 - x1) * 0.2)
            pad_y = int((y2 - y1) * 0.2)
            x1 = max(0, x1 - pad_x)
            y1 = max(0, y1 - pad_y)
            x2 = min(w, x2 + pad_x)
            y2 = min(h, y2 + pad_y)
            
            # Crop face (RGB)
            face_crop = cv2.cvtColor(frame[y1:y2, x1:x2], cv2.COLOR_BGR2RGB)
            
            if face_crop.size > 0:
                # Preprocess cropped face
                input_tensor = val_transform(face_crop).unsqueeze(0).to(device)
                
                # Predict liveness
                with torch.no_grad():
                    outputs = model(input_tensor)
                    probs_out = torch.softmax(outputs, dim=1)
                    pred_class = outputs.argmax(dim=1).item()
                    confidence = probs_out[0][pred_class].item() * 100
                
                # Draw results
                label = "REAL" if pred_class == 1 else "SPOOF"
                color = (0, 255, 0) if pred_class == 1 else (0, 0, 255)
                
                # Draw face box
                cv2.rectangle(frame, (x1, y1), (x2, y2), color, 2)
                
                # Draw label
                cv2.putText(frame, f"{label}: {confidence:.1f}%", (x1, y1-10), 
                            cv2.FONT_HERSHEY_SIMPLEX, 0.9, color, 2)
                cv2.putText(frame, f"Real: {probs_out[0][1].item()*100:.1f}%", (10, 30),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
                cv2.putText(frame, f"Spoof: {probs_out[0][0].item()*100:.1f}%", (10, 60),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
                cv2.putText(frame, f"Face: {face_conf*100:.1f}%", (10, 90),
                            cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
        else:
            cv2.putText(frame, "No face detected", (10, 30), 
                        cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 165, 255), 2)
        
        cv2.putText(frame, "FACE DETECTION + LIVENESS", (10, frame.shape[0]-10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 255), 1)
        
        cv2.imshow('Liveness Detection (Proper Pipeline)', frame)
        frame_count += 1
        
        if cv2.waitKey(1) & 0xFF == ord('q'):
            print("\n⏹️ Stopped by user")
            break
        if time.time() - start_time > duration:
            print(f"\n⏱️ Test completed ({duration}s)")
            break
    
    cap.release()
    cv2.destroyAllWindows()
    
    elapsed = time.time() - start_time
    fps = frame_count / elapsed
    print(f"\nProcessed {frame_count} frames in {elapsed:.1f}s ({fps:.1f} FPS)")
    print(f"Faces detected: {face_count}/{frame_count} ({face_count/frame_count*100:.1f}%)")
    print("=" * 60)

print("✅ Face detection + liveness functions defined")
print("👉 Run the next cell to test manually")
```

## 🔄 Retrain with Stronger Augmentation (Lighting Robustness)

The model works but is sensitive to lighting. Let's retrain with aggressive brightness/contrast augmentation to make it robust to different lighting conditions.

```python
# ============================================
# STRONGER AUGMENTATION FOR LIGHTING ROBUSTNESS
# ============================================
from torchvision import transforms
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, random_split

# Much stronger augmentation for lighting variations
robust_train_transform = transforms.Compose([
    transforms.ToPILImage(),
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(p=0.5),
    # STRONGER brightness/contrast augmentation
    transforms.ColorJitter(
        brightness=0.5,    # Was 0.2, now 0.5 (50% variation)
        contrast=0.5,      # Was 0.2, now 0.5
        saturation=0.3,    # Was 0.2, now 0.3
        hue=0.1            # Added hue variation
    ),
    transforms.RandomRotation(15),
    # Add random grayscale (simulates low light)
    transforms.RandomGrayscale(p=0.1),
    # Add Gaussian blur (simulates camera quality)
    transforms.GaussianBlur(kernel_size=3, sigma=(0.1, 2.0)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

print("✅ Robust transforms defined with:")
print("   - Brightness variation: ±50%")
print("   - Contrast variation: ±50%")
print("   - Random grayscale: 10%")
print("   - Gaussian blur for camera quality")
print("")
print("Run the next cell to retrain the model (~15-20 minutes)")
```

```python
# ============================================
# RETRAIN MODEL WITH ROBUST AUGMENTATION
# ============================================
import torch.optim as optim
from tqdm import tqdm

# Reload dataset with robust transforms
print("📂 Loading dataset with robust augmentation...")
robust_dataset = LivenessDataset(OULU_NPU_PATH, transform=robust_train_transform)

# Balance the dataset (undersample spoof class)
real_indices = [i for i, (_, label, _) in enumerate(robust_dataset.samples) if label == 1]
spoof_indices = [i for i, (_, label, _) in enumerate(robust_dataset.samples) if label == 0]

print(f"Real samples: {len(real_indices)}")
print(f"Spoof samples: {len(spoof_indices)}")

# Undersample spoof to match real
import random
random.seed(42)
spoof_indices_balanced = random.sample(spoof_indices, len(real_indices))
balanced_indices = real_indices + spoof_indices_balanced

# Create balanced subset
from torch.utils.data import Subset
balanced_dataset = Subset(robust_dataset, balanced_indices)
print(f"Balanced dataset: {len(balanced_dataset)} samples")

# Split into train/val
train_size = int(0.8 * len(balanced_dataset))
val_size = len(balanced_dataset) - train_size
train_dataset, val_dataset = random_split(balanced_dataset, [train_size, val_size])

# Create dataloaders
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, num_workers=0)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False, num_workers=0)

print(f"\nTrain: {len(train_dataset)} samples")
print(f"Val: {len(val_dataset)} samples")

# Create fresh model
model_robust = LivenessDetector(pretrained=True)
model_robust = model_robust.to(device)

# Stage 1: Train classifier only
model_robust.freeze_backbone()

print("\n" + "=" * 60)
print("STAGE 1: Training classifier (5 epochs)")
print("=" * 60)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model_robust.parameters(), lr=0.001, weight_decay=1e-4)

for epoch in range(5):
    model_robust.train()
    train_loss, correct, total = 0, 0, 0
    
    for images, labels in tqdm(train_loader, desc=f'Epoch {epoch+1}/5'):
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model_robust(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        
        train_loss += loss.item()
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()
    
    print(f"  Train Acc: {100.*correct/total:.2f}%")

# Stage 2: Fine-tune all layers
model_robust.unfreeze_backbone()
optimizer = optim.Adam(model_robust.parameters(), lr=0.0001, weight_decay=1e-4)

print("\n" + "=" * 60)
print("STAGE 2: Fine-tuning all layers (5 epochs)")
print("=" * 60)

best_acc = 0
for epoch in range(5):
    model_robust.train()
    for images, labels in tqdm(train_loader, desc=f'Epoch {epoch+1}/5'):
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model_robust(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
    
    # Validate
    model_robust.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for images, labels in val_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model_robust(images)
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()
    
    val_acc = 100. * correct / total
    print(f"  Val Acc: {val_acc:.2f}%")
    
    if val_acc > best_acc:
        best_acc = val_acc
        torch.save({
            'model_state_dict': model_robust.state_dict(),
            'val_acc': val_acc,
        }, '../models/liveness_robust.pth')
        print(f"  ✅ Saved best model")

print("\n" + "=" * 60)
print(f"TRAINING COMPLETE - Best Acc: {best_acc:.2f}%")
print("Model saved to: ../models/liveness_robust.pth")
print("=" * 60)
```

```python
# ============================================
# TEST ROBUST MODEL WITH WEBCAM
# ============================================
import cv2
import time

# Load the robust model
checkpoint = torch.load('../models/liveness_robust.pth', map_location=device)
model_robust.load_state_dict(checkpoint['model_state_dict'])
model_robust.eval()
print(f"✅ Loaded robust model (acc: {checkpoint['val_acc']:.2f}%)")

def test_robust_model(model, device, duration=30):
    """Test the lighting-robust model"""
    cap = cv2.VideoCapture(0)
    if not cap.isOpened():
        print("❌ Could not open webcam")
        return
    
    val_transform = transforms.Compose([
        transforms.ToPILImage(),
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ])
    
    print("=" * 60)
    print("ROBUST MODEL TEST (Lighting-Invariant)")
    print("=" * 60)
    print(f"Running for {duration} seconds... Press 'q' to quit")
    print("=" * 60)
    
    model.eval()
    start_time = time.time()
    
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        
        rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        input_tensor = val_transform(rgb_frame).unsqueeze(0).to(device)
        
        with torch.no_grad():
            outputs = model(input_tensor)
            probs = torch.softmax(outputs, dim=1)
            pred_class = outputs.argmax(dim=1).item()
            confidence = probs[0][pred_class].item() * 100
        
        label = "REAL" if pred_class == 1 else "SPOOF"
        color = (0, 255, 0) if pred_class == 1 else (0, 0, 255)
        
        cv2.putText(frame, f"{label}: {confidence:.1f}%", (10, 30), 
                    cv2.FONT_HERSHEY_SIMPLEX, 1, color, 2)
        cv2.putText(frame, f"Real: {probs[0][1].item()*100:.1f}%", (10, 70),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
        cv2.putText(frame, f"Spoof: {probs[0][0].item()*100:.1f}%", (10, 100),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 255), 2)
        cv2.putText(frame, "ROBUST MODEL", (10, 130),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 255), 1)
        
        cv2.imshow('Robust Liveness Detection', frame)
        
        if cv2.waitKey(1) & 0xFF == ord('q'):
            print("\n⏹️ Stopped by user")
            break
        if time.time() - start_time > duration:
            print(f"\n⏱️ Test completed ({duration}s)")
            break
    
    cap.release()
    cv2.destroyAllWindows()

print("\n👉 Run: test_robust_model(model_robust, device, 30)")
```

```python
# ============================================
# FINE-TUNE ON TAPAKAH68 (Real-world mobile captures)
# ============================================
# tapakah68 has real mobile phone selfies - closer to webcam input

print("📂 Loading tapakah68 with robust augmentation...")
tapakah_robust = LivenessDataset(TAPAKAH_PATH, transform=robust_train_transform)

# Create dataloader (small dataset, use all for training)
tapakah_loader = DataLoader(tapakah_robust, batch_size=8, shuffle=True, num_workers=0)

print(f"Tapakah68: {len(tapakah_robust)} samples")
print(f"  Real: {sum(tapakah_robust.labels)}")
print(f"  Spoof: {len(tapakah_robust.labels) - sum(tapakah_robust.labels)}")

# Fine-tune the robust model on tapakah68
print("\n" + "=" * 60)
print("FINE-TUNING ON TAPAKAH68 (3 epochs)")
print("=" * 60)

optimizer = optim.Adam(model_robust.parameters(), lr=0.00005, weight_decay=1e-4)

for epoch in range(3):
    model_robust.train()
    correct, total = 0, 0
    
    for images, labels in tqdm(tapakah_loader, desc=f'Epoch {epoch+1}/3'):
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model_robust(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += predicted.eq(labels).sum().item()
    
    print(f"  Acc: {100.*correct/total:.2f}%")

# Save the final model
torch.save({
    'model_state_dict': model_robust.state_dict(),
    'description': 'Robust + Tapakah68 fine-tuned'
}, '../models/liveness_final_robust.pth')

print("\n✅ Model saved to: ../models/liveness_final_robust.pth")
print("👉 Run the next cell to test")
```

```python
# ============================================
# TEST FINAL ROBUST MODEL
# ============================================
# Load and test the tapakah68 fine-tuned model
import torch
checkpoint = torch.load('../models/liveness_final_robust.pth', map_location=device)
model_robust.load_state_dict(checkpoint['model_state_dict'])
model_robust.eval()
print("✅ Loaded final robust model")

# Test with webcam
test_robust_model(model_robust, device, 30)
```
